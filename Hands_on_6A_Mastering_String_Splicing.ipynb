## Exercise 2

```python
original_log = "    LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25    "

print("Original Log:")
print(f'"{original_log}"\n')

cleaned_log = original_log.strip()

print("Cleaned Log:")
print(f"{cleaned_log}\n")

lowercase_log = cleaned_log.lower()

position = cleaned_log.find("SUCCESSFUL")

print("Position of SUCCESSFUL:")
print(f"{position}\n")

updated_log = cleaned_log.replace("SUCCESSFUL", "VERIFIED")

print("Updated Log:")
print(f"{updated_log}\n")
```

# Exercise 3

```python
movie_title = "The Last Kingdom Returns"

print("Movie Title:")
print(f"{movie_title}\n")

print("First Character:")
print(f"{movie_title[0]}\n")

print("Last Character:")
print(f"{movie_title[-1]}\n")

first_space = movie_title.find(" ")
print("First Word:")
print(f"{movie_title[:first_space]}\n")

last_space = movie_title.rfind(" ")
print("Last Word:")
print(f"{movie_title[last_space + 1:]}\n")

print('Position of "Returns":')
print(f'{movie_title.find("Returns")}\n')

print("Split Result:")
print(f"{movie_title.split()}\n")
```
