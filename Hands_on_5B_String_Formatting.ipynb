### String Formatting in Python


#### EXERCISE 1 : STUDENT REPORT CARD

```python
# Variables
student_name = "Esther Johnson"
age = 17
subject = "Mathematics"
score = 91.75

# Formatted Output
report_card = f"""==============================
        STUDENT REPORT
==============================

Student : {student_name}
Age     : {age} years
Subject : {subject}
Score   : {score:.2f}%

Congratulations {student_name}!
You passed {subject}.

=============================="""

print(report_card)
```

#### EXERCISE 2 : PAYROLL SYSTEM

```python
# Variables
emp_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 7.5

# Calculations
gross_salary = basic_salary + bonus
# Convert percentage to decimal for calculation (7.5 / 100)
tax_amount = gross_salary * (tax_rate / 100) 
net_salary = gross_salary - tax_amount

# Formatted Output
payroll_receipt = f"""==================================
        PAYROLL SUMMARY
==================================

Employee     : {emp_name}
Department   : {department}
Basic Salary : ₦{basic_salary:,.2f}
Bonus        : ₦{bonus:,.2f}
Gross Salary : ₦{gross_salary:,.2f}
Tax ({tax_rate}%)   : ₦{tax_amount:,.2f}
Net Salary   : ₦{net_salary:,.2f}

=================================="""

print(payroll_receipt)
```

#### EXERCISE 3 : E-COMMERCE RECEIPT

```python
# Variables
cust_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500

# Base Calculation
total_cost = quantity * unit_price

# Challenge: Apply 10% discount if quantity > 2
discount = 0
if quantity > 2:
    discount = total_cost * 0.10
    
final_amount = total_cost - discount

# Base Formatted Output
receipt = f"""====================================
      ONLINE STORE RECEIPT
====================================

Customer   : {cust_name}
Product    : {product}
Quantity   : {quantity}
Unit Price : ₦{unit_price:,.2f}
Total Cost : ₦{total_cost:,.2f}"""

# Conditionally append discount information if applicable
if discount > 0:
    receipt += f"""
Discount   : ₦{discount:,.2f}
Final Amount : ₦{final_amount:,.2f}"""

# Append the closing message
receipt += f"""

Thank you for shopping with us!
===================================="""

print(receipt)
```
