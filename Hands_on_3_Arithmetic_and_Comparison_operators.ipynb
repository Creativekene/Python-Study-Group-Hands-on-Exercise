```python

basic_salary = 180000  
performance_bonus = 45000  
eligibility_threshold = 220000  

total_salary = basic_salary + performance_bonus

is_eligible_for_premium = total_salary >= eligibility_threshold

print(f"Total Salary: N{total_salary:,}")
print(f"Eligible for Premium Benefits: {is_eligible_for_premium}")
```

```python

python_score = 72
sql_score = 65
excel_score = 80
powerbi_score = 75
passing_threshold = 70

total_score = python_score + sql_score + excel_score + powerbi_score

number_of_assessments = 4
average_score = total_score / number_of_assessments

has_passed = average_score >= passing_threshold

print(f"Total Score: {total_score}")
print(f"Average Score: {average_score}")
print(f"Passed: {has_passed}")
```

```python

initial_stock = 250  # units
sold = 85  # units
reorder_threshold = 200  # units

remaining_stock = initial_stock - sold

reorder_needed = remaining_stock < reorder_threshold

print(f"Remaining Stock: {remaining_stock}")
print(f"Reorder Needed: {reorder_needed}")
```

QUESTION 4

```python

laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000
discount_threshold = 50000
discount_rate = 0.10 

total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive


discount_qualified = total_purchase >= discount_threshold


if discount_qualified:
    discount_amount = total_purchase * discount_rate
else:
    discount_amount = 0

final_amount = total_purchase - discount_amount


print(f"Total Purchase: N{total_purchase:,}")
print(f"Discount Qualified: {discount_qualified}")
print(f"Discount Amount: N{discount_amount:,}")
print(f"Final Amount: N{final_amount:,}")
```
