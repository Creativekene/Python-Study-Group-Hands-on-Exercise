## STUDENT SCORE TRACKER

```python
# 1. Create variable with initial value
score = 60

# 2. Add 15 marks for class assignment 
score += 15

# 3. Subtract 5 marks for late submission 
score -= 5

# 4. Double the score for special bonus
score *= 2

# 5. Print the final score
print("Final Score:", score)
```

## DIGITAL WALLET SIMULATOR

```python
# 1. Create variable with initial balance
balance = 25000

# 2. Deposit 8500
balance += 8500

# 3. Withdraw 4200
balance -= 4200

# 4. Divide balance equally among 3 savings accounts
balance /= 3

# 5. Print the final balance
print("Final Balance in each account: ₦", round(balance, 2))

# 6. Print the data type of the balance
print("Data type:", type(balance))
```

## INVENTORY MANAGEMENT SYSTEM

```python
# 1. Create variable with initial stock
stock = 150

# 2. Customer buys 28 items
stock -= 28

# 3. New shipment of 75 items arrives
stock += 75

# 4. Pack remaining items into cartons of 12 
stock //= 12

# 5. Print the number of full cartons
print("Number of full cartons:", stock)
```
